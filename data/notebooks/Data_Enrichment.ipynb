```python
categories = list(product_catalog.keys())

product_master = []

for i, sku in enumerate(products["SKU_ID"]):

    category = categories[i % len(categories)]

    brand = product_catalog[category]["brands"][
        i % len(product_catalog[category]["brands"])
    ]

    product = product_catalog[category]["products"][
        i % len(product_catalog[category]["products"])
    ]

    product_master.append([
        sku,
        f"{brand} {product}",
        brand,
        category
    ])

product_master = pd.DataFrame(
    product_master,
    columns=[
        "SKU_ID",
        "Product_Name",
        "Brand",
        "Category"
    ]
)

product_master.head(10)
```

```python
duplicates = product_master["Product_Name"].duplicated(keep=False)

product_master.loc[duplicates, "Product_Name"] = (
    product_master.loc[duplicates, "Product_Name"]
    + " "
    + product_master.loc[duplicates, "SKU_ID"]
)
```

## Conclusion drawn after running data enrichment on our Original dataset

```python
# Since we got :
# WH_1 → 4 regions
# WH_2 → 4 regions
# WH_3 → 4 regions
# WH_4 → 4 regions
# WH_5 → 4 regions

# Warehouse_ID → Region
# is NOT a valid functional dependency.

# Therefore, we cannot create:

# Warehouses
# '-----------'
# Warehouse_ID
# Region

# and assume one region per warehouse.

# This tells us something interesting about the dataset: Region behaves like a transaction-level attribute rather than a fixed warehouse attribute.
```

```python
# Since we got :
# SUP_1     12
# SUP_2     11
# SUP_3     12
# ...
# SUP_8     13

# So a supplier has multiple lead times.

# Therefore:

# Supplier_ID → Supplier_Lead_Time_Days
# is also NOT valid.

# We shouldn't create a Suppliers table with a single lead time per supplier.
```

```python
# Every SKU has:
# 5 different Unit_Cost values

# Therefore:

# SKU_ID → Unit_Cost

# is invalid.
```

```python
# Same situation:

# SKU_ID → Unit_Price

# is invalid.
```

```python
# Again:

# SKU_ID → Reorder_Point

# is invalid.
```

```python
# We found out :
# sku_warehouse_reorder.max()
#         ↓
#         1

# So:

# SKU_ID + Warehouse_ID
#               ↓
#        Reorder_Point

# is a valid dependency based on the current test.

# Therfore: the reorder point belongs to an inventory-level relationship rather than the product master.
```

```python
# Stockout_Flag = 0 for all 91,250 records.

# Therefore we should not use stockout rate as a project KPI.

# We'll instead use things like:

# Inventory < Reorder Point

# for inventory-risk analysis.
```
